```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import BernoulliNB, GaussianNB, MultinomialNB
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn import metrics
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("city_day.csv")
data.head()
```

```python
num_cols = data.select_dtypes(include=['float64','int64']).columns
data[num_cols] = data[num_cols].fillna(data[num_cols].mean())

obj_cols = data.select_dtypes(include=['object']).columns
for c in obj_cols:
    data[c] = data[c].fillna(data[c].mode()[0])
```

```python
x = data.drop("AQI_Bucket", axis=1)
y = data["AQI_Bucket"]

```

```python
cat_cols = x.select_dtypes(['object']).columns
x[cat_cols] = x[cat_cols].apply(lambda x: pd.factorize(x)[0])

y = pd.factorize(y)[0]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    x, y, test_size=0.3, shuffle=False
)
```

```python
RF = RandomForestClassifier()
RF.fit(X_train, Y_train)
Y_rpred = RF.predict(X_test)
r_accuracy = metrics.accuracy_score(Y_test, Y_rpred)
r_accuracy
```

```python
BernNB = BernoulliNB()
BernNB.fit(X_train, Y_train)
Y_bpred = BernNB.predict(X_test)
b_accuracy = metrics.accuracy_score(Y_test, Y_bpred)
b_accuracy
```

```python
GausNB = GaussianNB()
GausNB.fit(X_train, Y_train)
Y_gpred = GausNB.predict(X_test)
g_accuracy = metrics.accuracy_score(Y_test, Y_gpred)
g_accuracy
```

```python
Dtree = DecisionTreeClassifier()
Dtree.fit(X_train, Y_train)
Y_dpred = Dtree.predict(X_test)
d_accuracy = metrics.accuracy_score(Y_test, Y_dpred)
d_accuracy
```

```python
MultiNB = MultinomialNB()
MultiNB.fit(X_train, Y_train)
Y_mpred = MultiNB.predict(X_test)
m_accuracy = metrics.accuracy_score(Y_test, Y_mpred)
m_accuracy
```

```python
KNN = KNeighborsClassifier()
KNN.fit(X_train, Y_train)
Y_kpred = KNN.predict(X_test)
k_accuracy = metrics.accuracy_score(Y_test, Y_kpred)
k_accuracy
```

```python
print("Bernoulli:", b_accuracy)
print("Random Forest:", r_accuracy)
print("Gaussian:", g_accuracy)
print("Decision Tree:", d_accuracy)
print("Multinomial:", m_accuracy)
print("KNN:", k_accuracy)
```
